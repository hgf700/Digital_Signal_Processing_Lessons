# Encode and decode the signal [15, 20, 25, 25, 30] using delta coding.

```python
import numpy as np

def delta_encode(signal):
    return np.diff(np.insert(signal, 0, signal[0]))

def delta_decode(encoded_signal, start_value):
    return start_value + np.cumsum(encoded_signal)

original_signal = [15, 20, 25, 25, 30]

encoded_signal = delta_encode(original_signal)
decoded_signal = delta_decode(encoded_signal, original_signal[0])

print("Original Signal:", original_signal)
print("Encoded Signal:", encoded_signal)
print("Decoded Signal:", decoded_signal)
```
