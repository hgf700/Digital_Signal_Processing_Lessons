# Analyze a sine wave with frequency f = 10 Hz, sampled at fs = 15 Hz.

```python
from scipy.signal import resample
import numpy as np
import matplotlib.pyplot as plt

# Reconstructing the signal using high sampling rate
f_signal = 10

t = np.linspace(0, 1, 1000, endpoint=False)  
signal = np.sin(2 * np.pi * f_signal * t)  

f_sample=15
t_sample = np.arange(0, 1, 1 / f_sample)
samples = np.sin(2 * np.pi * f_signal * t_sample)

num_samples = 1000
reconstructed_signal = resample(samples, num_samples)

plt.plot(t, signal, label='Original Signal')
plt.stem(t_sample, samples, linefmt='r-', markerfmt='ro', basefmt=" ", label='Samples (15 Hz)')
plt.plot(t, reconstructed_signal, label='Reconstructed Signal', linestyle='--')
plt.title('Signal Reconstruction (fs = 15 Hz)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.legend()
plt.grid()
plt.show()
```

# Encode and decode the signal [15, 20, 25, 25, 30] using delta coding.

```python
import numpy as np

def delta_encode(signal):
    return np.diff(np.insert(signal, 0, signal[0]))

def delta_decode(encoded_signal, start_value):
    return start_value + np.cumsum(encoded_signal)

original_signal = [15, 20, 25, 25, 30]

encoded_signal = delta_encode(original_signal)
decoded_signal = delta_decode(encoded_signal, original_signal[0])

print("Original Signal:", original_signal)
print("Encoded Signal:", encoded_signal)
print("Decoded Signal:", decoded_signal)
```
